```python
import pandas as pd
import glob

# Search for all CSV files in current folder
all_files = glob.glob("*.csv")

# Read each CSV and add it into a list
data_frames = []
for file in all_files:
    df_temp = pd.read_csv(file)
    data_frames.append(df_temp)

# Combine all CSVs into one single dataframe
df = pd.concat(data_frames, ignore_index=True)

# Print shape and column names
print("Total rows & columns loaded:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

# Show first 5 rows
df.head()
```

```python
# Check dataset info and missing values
print(df.info())

# Show column names clearly
print("\nYour Columns:")
print(list(df.columns))
```

```python
from sklearn.preprocessing import StandardScaler

# Select only numerical sensor columns for K-Means
features = ['acc_x', 'acc_y', 'acc_z', 'gyro_x', 'gyro_y', 'gyro_z', 'mag_x', 'mag_y', 'mag_z']
X = df[features]

# K-Means-er jonno data scale (standardize) kora zaroori
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Data Scaling Done! Shape:", X_scaled.shape)
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Elbow Method to find optimal k
wcss = []
for k in range(1, 10):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plot the Elbow Graph
plt.figure(figsize=(8, 5))
plt.plot(range(1, 10), wcss, marker='o', linestyle='--', color='b')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('WCSS (Inertia)')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
from sklearn.cluster import KMeans

# 1. Train K-Means with k=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)

# 2. Visualize the Clusters using Accelerometer X and Y
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='acc_x',
    y='acc_y',
    hue='Cluster',
    data=df,
    palette='Set1',
    alpha=0.6
)
plt.title('Driving Behavior Clusters (K-Means with k=3)')
plt.xlabel('Acceleration X (acc_x)')
plt.ylabel('Acceleration Y (acc_y)')
plt.grid(True)
plt.show()

# 3. Print mean sensor values for each cluster to understand driving behavior
print("\nAverage Sensor Values per Cluster:")
print(df.groupby('Cluster')[['acc_x', 'acc_y', 'acc_z', 'gyro_x', 'gyro_y', 'gyro_z']].mean())
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import adjusted_rand_score, cluster
from scipy.optimize import linear_sum_assignment

# 1. Compute Contingency Matrix (Compare K-Means Clusters with Real Harsh Events)
cross_tab = pd.crosstab(
    df['Cluster'].map(lambda c: f'Cluster {c}'),
    df['harsh_event'],
    rownames=['K-Means Cluster'],
    colnames=['True Harsh Event']
)

# 2. Add diagnostic metrics: dominant event and purity
cross_tab['Cluster Size'] = cross_tab.sum(axis=1)
cross_tab['Dominant Event'] = cross_tab.drop(columns='Cluster Size').idxmax(axis=1)
cross_tab['Purity (%)'] = (
    cross_tab.drop(columns=['Cluster Size', 'Dominant Event']).max(axis=1) / cross_tab['Cluster Size'] * 100
).round(2)

print("--- Contingency Matrix with Cluster Diagnostics ---")
display(cross_tab)

# 3. Compute Adjusted Rand Index (ARI)
ari_score = adjusted_rand_score(df['harsh_event'], df['Cluster'])
print(f"\nAdjusted Rand Index (ARI): {ari_score:.4f}")

# 4. Heatmap Visualization
plt.figure(figsize=(10, 6))
sns.heatmap(cross_tab.drop(columns=['Cluster Size', 'Dominant Event', 'Purity (%)']), annot=True, fmt='d', cmap='Blues')
plt.title(f"Contingency Heatmap (K-Means vs True Harsh Events)\nARI Score: {ari_score:.4f}", fontsize=12, fontweight='bold')
plt.xlabel("True Harsh Event")
plt.ylabel("K-Means Predicted Cluster")
plt.tight_layout()
plt.show()
```
